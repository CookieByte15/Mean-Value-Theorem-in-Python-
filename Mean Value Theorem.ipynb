```python
import math

def f(expr, x):
    return eval(expr, {"x": x, "math": math})


def mean_value_theorem_root(expr, a, b, tol=1e-5, max_iter=1000):
    fa = f(expr, a)
    fb = f(expr, b)

    # Endpoints do not confirm a root in the range
    if fa * fb > 0:
        return None

    for _ in range(max_iter):
        mid = (a + b) / 2
        fmid = f(expr, mid)

        if abs(fmid) < tol:
            return mid

        if fa * fmid < 0:
            b = mid
            fb = fmid
        else:
            a = mid
            fa = fmid

        if abs(b - a) < tol:
            return (a + b) / 2

    return (a + b) / 2


expr = input("Enter the equation in terms of x: ")
a = float(input("Enter the start of the interval (a): "))
b = float(input("Enter the end of the interval (b): "))

root = mean_value_theorem_root(expr, a, b)

if root is None:
    print(
        "No root confirmed in the interval [{}, {}] "
        "based on endpoint signs.".format(a, b)
    )
else:
    print("Root found at approximately x =", root)
```
